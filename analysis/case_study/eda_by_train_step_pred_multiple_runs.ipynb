```python
from utils_analysis_by_step_pred import (
    load_prediction_results,
    display_preds_for_idx,
    generate_vocabs_any_step,
    generate_vocabs_final_step,
    count_num_sents_with_word,
    find_idxs_where_present,
    generate_vocab_lists,
    load_pickle,
    extract_predictions,
)


folder = "eng_fil"
language_code = "tl"
filename = "ordering_results_pred_by_step_bs_1_1eps.json"
```

```python
def count_languages_from_vocabs(detected_langs: list[list[tuple[str, str]]]):
    detected_lang_counts = dict[str, int]()
    for item in detected_langs:
        for lang in set([word[1] for word in item[1]]):
            if lang in detected_lang_counts:
                detected_lang_counts[lang] += 1
            else:
                detected_lang_counts[lang] = 1
    return sorted(detected_lang_counts.items(), key=lambda item: -item[1])
```

```python
cache = load_pickle(f"assets/{folder}_detected_words.pkl")
```

```python
predictions_run1 = load_prediction_results(folder, filename, run=0)
predictions_run2 = load_prediction_results(folder, filename, run=1)
predictions_run3 = load_prediction_results(folder, filename, run=2)
predictions_run4 = load_prediction_results(folder, filename, run=3)
predictions_run5 = load_prediction_results(folder, filename, run=4)
```

```python
train_vocabs, test_vocabs = await generate_vocab_lists(
    predictions_run1["train_targets"],
    predictions_run1["test_targets"],
    folder,
    language_code,
)
```

```python
outputs_run1 = await extract_predictions(
    predictions=predictions_run1,
    train_vocabs=list(train_vocabs),
    test_vocabs=list(test_vocabs),
    cache=cache,
)

outputs_run2 = await extract_predictions(
    predictions=predictions_run2,
    train_vocabs=list(train_vocabs),
    test_vocabs=list(test_vocabs),
    cache=cache,
)

outputs_run3 = await extract_predictions(
    predictions=predictions_run3,
    train_vocabs=list(train_vocabs),
    test_vocabs=list(test_vocabs),
    cache=cache,
)
outputs_run4 = await extract_predictions(
    predictions=predictions_run4,
    train_vocabs=list(train_vocabs),
    test_vocabs=list(test_vocabs),
    cache=cache,
)
outputs_run5 = await extract_predictions(
    predictions=predictions_run5,
    train_vocabs=list(train_vocabs),
    test_vocabs=list(test_vocabs),
    cache=cache,
)
```
